```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
sns.load_dataset('penguins')
```

```python
df = sns.load_dataset('penguins').dropna()
df
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x='island', data=df,palette="Set2")
plt.title("Q1: Area Marks showing counts")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(x='bill_length_mm',
                y="body_mass_g",
                hue="species",
                size="flipper_length_mm",
                data=df,
                sizes=(20,200),
                alpha=0.7)
plt.title("Q2: Area Marks showing counts")
plt.show()
```
